Let's start by creating a simple image manually

```python
A = np.zeros((11,11), dtype=np.uint8)
A[2, 2:9] = 255
A[3:8, [2,8]] = 255
A[8, 2:9] = 255
plt.imshow(A, cmap="gray")
plt.axis("off")
```

Let's define the convolution operation where the filter is applied to each pixel

```python
def convolve2d(img, kernel):
    print("Original image", img.shape)
    print(img)
    h, w = img.shape
    kh, kw = kernel.shape
    pad_h, pad_w = kh//2, kw//2

    padded = np.pad(img, ((pad_h, pad_h), (pad_w, pad_w)), mode="constant", constant_values=(0))
    out = np.zeros_like(img, dtype=np.float32)
    print("After padding", padded.shape)
    print(padded)

    for i in range(h):
        for j in range(w):
            region = padded[i:i+kh, j:j+kw]
            out[i, j] = np.sum(region * kernel)
    print("After convolution", out.shape)
    print(out)
    out = np.abs(((out - out.min()) / (out.max()-out.min()) * 255)).astype(np.uint8)
    # out = np.abs(out).astype(np.uint8)
    print("After renormalization of values into 0-255 range", out.shape)
    print(out)
    return out
```

## Edge detection

```python
fv = np.array([[-1, 0, 1],
               [-1, 0, 1],
               [-1, 0, 1]])

ver_edges = convolve2d(A, fv)
plt.imshow(ver_edges, cmap="gray")
plt.axis("off")
```

Notice how there is a bright edge whenever you move from dark to bright regions and a dark edge whenever you move from bright to dark regions.

Let's try the same operation on the chessboard image

```python
img_ver_edges = convolve2d(img_arr, fv)
plt.imshow(img_ver_edges, cmap="gray")
plt.axis("off")
```

# Rotations

### Clockwise rotation by 90 degrees
Let's begin with rotating the image clockwise by 90 degrees

```python
def rotate_clockwise_90_degrees(img):
  m, n = img.shape
  last_row = m - 1 # Since numbering in python arrays begins from 0 and for to m-1
  last_col = n - 1 # Since numbering in python arrays begins from 0 and for to m-1
  rotated_img = np.zeros((n,m))
  for i in range(0, m): # This loop runs from 0 to m-1, since the lower value is included but the upper value is not included by the range function
    for j in range(0, n): # This loop runs from 0 to n-1, since the lower value is included but the upper value is not included by the range function
      rotated_img[j, last_row-i] = img[i, j]
  return rotated_img
```

```python
clockwise_rotated_img_90 = rotate_clockwise_90_degrees(img_arr)
plt.imshow(clockwise_rotated_img_90, cmap="gray")
plt.axis("off")
```
